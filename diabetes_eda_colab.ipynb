## Diabetes Data Quick Check
- Goal: look at the dataset and see simple patterns.
- I kept this notebook short and focused on basic EDA.


```python
from google.colab import files

uploaded = files.upload()
print('Uploaded:', list(uploaded.keys()))
```

### Load data
- First I import the libraries and read the CSV file.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style='whitegrid')

file_name = list(uploaded.keys())[0]
df = pd.read_csv(file_name)
df.head()
```

### First look
- I checked the size, column names, and data types.


```python
print('Shape:', df.shape)
print('Columns:', list(df.columns))
print()
df.info()
```

### Missing and zero values
- I wanted to see if the file has missing values.
- I also checked zero values in medical columns because some of them may be unrealistic.


```python
zero_cols = ['Glucose', 'BloodPressure', 'SkinThickness', 'Insulin', 'BMI']

missing_values = df.isnull().sum().rename('missing_count')
zero_values = (df[zero_cols] == 0).sum().rename('zero_count')

display(missing_values.to_frame())
display(zero_values.to_frame())
```

### Summary statistics
- This gives a quick numeric summary of the dataset.


```python
df.describe().T
```

### Small NumPy check
- I used NumPy to calculate two simple values.


```python
glucose_median = np.median(df['Glucose'])
bmi_75 = np.percentile(df['BMI'], 75)

print(f'Glucose median: {glucose_median:.1f}')
print(f'BMI 75th percentile: {bmi_75:.1f}')
```

### Simple plots
- I checked class balance and the distribution of Glucose and BMI.
- I also compared Glucose by Outcome.


```python
fig, axes = plt.subplots(2, 2, figsize=(12, 8))

sns.countplot(data=df, x='Outcome', ax=axes[0, 0])
axes[0, 0].set_title('Outcome Count')

sns.histplot(data=df, x='Glucose', bins=20, ax=axes[0, 1], color='skyblue')
axes[0, 1].set_title('Glucose Distribution')

sns.histplot(data=df, x='BMI', bins=20, ax=axes[1, 0], color='salmon')
axes[1, 0].set_title('BMI Distribution')

sns.boxplot(data=df, x='Outcome', y='Glucose', ax=axes[1, 1])
axes[1, 1].set_title('Glucose by Outcome')

plt.tight_layout()
plt.show()
```

### Short findings
- The dataset has 768 rows and 9 columns.
- There are more `Outcome = 0` cases than `Outcome = 1`.
- I noticed many zero values in `Insulin` and `SkinThickness`.
- `Glucose` looks higher in the positive outcome group.
- I did not clean or change the data here. This is only a first check.
